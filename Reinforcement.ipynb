```python
import random
from typing import List
```

```python
class SampleEnvironment:
    def __init__(self):
        self.steps_left=20


    def get_actions(self) -> List[str]:
        return [0,1]

    def is_done(self):
        return self.steps_left == 0

    def actions(self, action: str):
        if self.is_done():
            raise Exception("Environment is done. No more actions can be taken.")
        self.steps_left -= 1
        if action==1:
            return random.random()+0.5
        else:
            return random.random()

    def get_reward(self, state: str, action: str) -> float:
        # Simulate a reward based on the action taken
        return random.uniform(-1, 1)
```

```python
class Agent:
    def __init__(self):
        self.total_reward = 0.0
    def step(self, env: SampleEnvironment):
        if env.is_done():
            print("Environment is done. No more actions can be taken.")
            return
        actions = env.get_actions()
        action = random.choice(actions)
        reward = env.actions(action)
        self.total_reward += reward
        print(f"Action taken: {action}, Reward received: {reward}, Total reward: {self.total_reward}")
```

```python
if __name__ == "__main__":
    env = SampleEnvironment()
    agent = Agent()
    while not env.is_done():
        agent.step(env)
    print(f"Final total reward: {agent.total_reward}")
```
